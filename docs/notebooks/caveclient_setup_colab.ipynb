# Setup a persistent token for CAVE authentication

**This is written for a Google Colab notebook.** If you are running this tutorial locally with Jupyter or VScode, consider the alternative notebook: [Local CAVEclient Setup](https://alleninstitute.github.io/Teaching_Connectomics/notebooks/caveclient-setup-local.html)

The **Connectome Annotation Versioning Engine (CAVE)** is a suite of tools developed at the Allen Institute and Seung Lab to manage large connectomics data.  

The CAVEclient is the main way to programmatically access the MICrONS data using Python. In particular, the CAVEclient provides an interface to query the CAVE database for annotations such as synapses, as well as to get a variety of other kinds of information.  

This notebook adds your Connectome Annotation Versioning Engine (CAVE) access token to you Google Colab 'Secrets' panel. Once you have setup a CAVE token in this way, you can access CAVE in any other colab notebook with the following:

```
from google.colab import userdata
import caveclient
client = caveclient.CAVEclient('minnie65_public',
                               auth_token=userdata.get('CAVE_TOKEN'))
```

The first time you run a notebook, you will be prompted the to grant it access to you 'Secrets'

````python
!uv pip install caveclient

from caveclient import CAVEclient
from google.colab import userdata

datastack_name = "minnie65_public"
server_address='https://global.daf-apis.com/'
````

# 1. Set up your CAVE access
1. Check to ensure you have accepted [MICrONS Terms of Service](https://global.daf-apis.com/sticky_auth/api/v1/tos/2/accept)
2. Visit <https://global.daf-apis.com//sticky_auth/settings/tokens>
3. Copy a "token" if one exists, or create a new one (upper right corner) if none are listed. Just copy the long alphanumeric part, not the quote marks.
4. Click the Key icon along the left of the page. Click + Add new secret.

![Access Colab 'Secrets' panel and set your CAVE_TOKEN](https://github.com/AllenInstitute/Teaching_Connectomics/blob/main/docs/resources/colab_set_token.png?raw=1)

5. Set the Name of the token to CAVE_TOKEN and paste the token as the Value.
6. Toggle "Notebook access" along the left side to be on.

# 2. Confirm CAVE access
Run the following cell. Click the link that is shown, and click through with your Google login credentials that you used previously when generating your token. You may need to accept a terms of service for the dataset.


````python
gclient = CAVEclient(server_address=server_address, auth_token=userdata.get('CAVE_TOKEN'), global_only=True)
info_url = f"{gclient.server_address}info/datastack/{datastack_name}"

client = CAVEclient(
  datastack_name=datastack_name,
  server_address=server_address,
  auth_token=userdata.get('CAVE_TOKEN'),
)

if isinstance(client.info.get_datastack_info(), dict):
  print("All set up! You can also access:\n", info_url)
````

If you you see a page loading that shows "Datastack:" a description and several service links, you should be all set up!

If you get a `HTTPError: 401 Client Error: UNAUTHORIZED for url` error, re-enter your token in Step 5 above
